```python
import os
import numpy as np
import pandas as pd
import shutil
from pathlib import Path

```

```python
data_dir = Path('../griffin_datasets/tel-griffin-v3')
griffin_data_dir = Path('../griffin_datasets/tel-pk')

```

```python
severity_type = pd.read_parquet(data_dir / 'severity.pqt')
severity_type['primary_key'] = severity_type.index
severity_type
```

```python
event_type = pd.read_parquet(data_dir / 'event.pqt')
event_type['primary_key'] = event_type.index
event_type

```

```python
resource_type = pd.read_parquet(data_dir / 'resource.pqt')
resource_type['primary_key'] = resource_type.index
resource_type

```

```python
log_feature = pd.read_parquet(data_dir / 'feature.pqt')
log_feature['primary_key'] = log_feature.index
log_feature


```

```python
# Saving the files
shutil.copytree(data_dir, griffin_data_dir)
severity_type.to_parquet(griffin_data_dir / 'severity.pqt')
event_type.to_parquet(griffin_data_dir / 'event.pqt')
resource_type.to_parquet(griffin_data_dir / 'resource.pqt')
log_feature.to_parquet(griffin_data_dir / 'feature.pqt')
```
