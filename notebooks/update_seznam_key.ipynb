```python
import numpy as np
import pandas as pd
```

```python
dataset_dir = '../datasets/seznam'
```

```python
client = pd.read_parquet(f'{dataset_dir}/data/client.pqt')
dobito = pd.read_parquet(f'{dataset_dir}/data/dobito.pqt')
probehnuto_mimo_penezenku = pd.read_parquet(f'{dataset_dir}/data/probehnuto_mimo_penezenku.pqt')
probehnuto = pd.read_parquet(f'{dataset_dir}/data/probehnuto.pqt')
```

```python
dobito
```

```python
# generate a mapping from dobito.index to a unique id
dobito_id = np.arange(len(dobito))
dobito_id_map = pd.Series(dobito_id, index=dobito.index)
dobito['dobito_id'] = dobito_id_map
```

```python
dobito
```

```python
charge_train = pd.read_parquet(f'{dataset_dir}/charge/train.pqt')
charge_validation = pd.read_parquet(f'{dataset_dir}/charge/validation.pqt')
charge_test = pd.read_parquet(f'{dataset_dir}/charge/test.pqt')
```

```python
charge_train
```

```python
# update charge train
charge_train['dobito_id'] = charge_train.index.map(dobito_id_map)

# update charge validation
charge_validation['dobito_id'] = charge_validation.index.map(dobito_id_map)

# update charge test
charge_test['dobito_id'] = charge_test.index.map(dobito_id_map)
```

```python
charge_train
```

```python
probehnuto = pd.read_parquet(f'{dataset_dir}/data/probehnuto.pqt')
probehnuto
```

```python
probehnuto_id = np.arange(len(probehnuto))
probehnuto_id_map = pd.Series(probehnuto_id, index=probehnuto.index)
probehnuto['probehnuto_id'] = probehnuto_id_map

probehnuto
```

```python
prepay_train = pd.read_parquet(f'{dataset_dir}/prepay/train.pqt')
prepay_validation = pd.read_parquet(f'{dataset_dir}/prepay/validation.pqt')
prepay_test = pd.read_parquet(f'{dataset_dir}/prepay/test.pqt')
```

```python
prepay_train['probehnuto_id'] = prepay_train.index.map(probehnuto_id_map)
prepay_validation['probehnuto_id'] = prepay_validation.index.map(probehnuto_id_map)
prepay_test['probehnuto_id'] = prepay_test.index.map(probehnuto_id_map)

prepay_train
prepay_validation
prepay_test
```

```python
probehnuto_mimo_penezenku = pd.read_parquet(f'{dataset_dir}/data/probehnuto_mimo_penezenku.pqt')
```

```python
probehnuto_mimo_penezenku
```

```python
probehnuto_mimo_penezenku_id = np.arange(len(probehnuto_mimo_penezenku))
probehnuto_mimo_penezenku_id_map = pd.Series(probehnuto_mimo_penezenku_id, index=probehnuto_mimo_penezenku.index)
probehnuto_mimo_penezenku['probehnuto_mimo_penezenku_id'] = probehnuto_mimo_penezenku_id_map

probehnuto_mimo_penezenku
```

```python
# Save the parquet files under griffin_dataset
griffin_dataset_path = '../griffin_datasets/seznam-griffin-pk'
dataset_path = dataset_dir
import shutil
import os
shutil.rmtree(griffin_dataset_path, ignore_errors=True)
shutil.copytree(dataset_path, griffin_dataset_path)
```

```python
# resave the parquet files
dobito.to_parquet(f'{griffin_dataset_path}/data/dobito.pqt')
probehnuto.to_parquet(f'{griffin_dataset_path}/data/probehnuto.pqt')
probehnuto_mimo_penezenku.to_parquet(f'{griffin_dataset_path}/data/probehnuto_mimo_penezenku.pqt')
charge_train.to_parquet(f'{griffin_dataset_path}/charge/train.pqt')
charge_validation.to_parquet(f'{griffin_dataset_path}/charge/validation.pqt')
charge_test.to_parquet(f'{griffin_dataset_path}/charge/test.pqt')
prepay_train.to_parquet(f'{griffin_dataset_path}/prepay/train.pqt')
prepay_validation.to_parquet(f'{griffin_dataset_path}/prepay/validation.pqt')
prepay_test.to_parquet(f'{griffin_dataset_path}/prepay/test.pqt')
```
