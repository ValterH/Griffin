```python
import numpy as np
import pandas as pd
```

```python
dataset_path = "../datasets/retailrocket"
griffin_dataset_path = "../griffin_datasets/retailrocket-pk"
```

```python
all_views = pd.read_parquet(f"{dataset_path}/data/all_views.pqt")
```

```python
pd.unique(all_views['added_to_cart'])
```

```python
all_views
```

```python
# Add the id as a key to the data
all_views["viewid"] = all_views.index
```

```python
all_views["added_to_cart"] = all_views["added_to_cart"].astype(bool)
all_views
```

```python
train_data = pd.read_parquet(f"{dataset_path}/cvr/cvr-100k_train.pqt")
validation_data = pd.read_parquet(f"{dataset_path}/cvr/cvr-100k_validation.pqt")
test_data = pd.read_parquet(f"{dataset_path}/cvr/cvr-100k_test.pqt")
```

```python
# find the row with viewid = 1364615
all_views[all_views["viewid"] == 1364615]
```

```python
train_data
```

```python
train_data["viewid"] = train_data.index
validation_data["viewid"] = validation_data.index
test_data["viewid"] = test_data.index
```

```python
train_data["added_to_cart"] = train_data["added_to_cart"].astype(bool)
validation_data["added_to_cart"] = validation_data["added_to_cart"].astype(bool)
test_data["added_to_cart"] = test_data["added_to_cart"].astype(bool)
```

```python
test_data["added_to_cart"].value_counts()
```

```python
test_data.head()
```

```python
item_availability = pd.read_parquet(f"{dataset_path}/data/item_availability.pqt")
item_availability["item_availability_id"] = np.arange(len(item_availability))
item_availability
```

```python
item_categories = pd.read_parquet(f"{dataset_path}/data/item_categories.pqt")
item_categories["item_categories_id"] = np.arange(len(item_categories))
item_categories
```

```python
item_property = pd.read_parquet(f"{dataset_path}/data/item_properties.pqt")
item_property["item_property_id"] = np.arange(len(item_property))
item_property
```

```python
# Save the parquet files under griffin_dataset
import shutil
import os
shutil.rmtree(griffin_dataset_path, ignore_errors=True)
shutil.copytree(dataset_path, griffin_dataset_path)
```

```python
# remove the cvr folder
shutil.rmtree(f"{griffin_dataset_path}/cvr", ignore_errors=True)
os.makedirs(f"{griffin_dataset_path}/cvr", exist_ok=True)
# Save the parquet files
train_data.to_parquet(f"{griffin_dataset_path}/cvr/cvr-100k_train.pqt")
validation_data.to_parquet(f"{griffin_dataset_path}/cvr/cvr-100k_validation.pqt")
test_data.to_parquet(f"{griffin_dataset_path}/cvr/cvr-100k_test.pqt")
all_views.to_parquet(f"{griffin_dataset_path}/data/all_views.pqt")

item_availability.to_parquet(f"{griffin_dataset_path}/data/item_availability.pqt")
item_categories.to_parquet(f"{griffin_dataset_path}/data/item_categories.pqt")
item_property.to_parquet(f"{griffin_dataset_path}/data/item_properties.pqt")
```
