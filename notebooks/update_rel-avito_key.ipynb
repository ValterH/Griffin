```python
import pandas as pd
import numpy as np

dataset_path = "../griffin_datasets/rel-avito"
griffin_path = "../griffin_datasets/rel-avito-pk"

```

```python

location = pd.read_parquet(f"{dataset_path}/Location.pqt")
# turn region_id and city_id to category, by
# 1. step 1: turning the nan to None
# 2. step 2: turning the left float to int
for col in ["RegionID", "CityID"]:
    location[col] = location[col].astype("Int64")
location
```

```python
SearchInfo = pd.read_parquet(f"{dataset_path}/SearchInfo.pqt")
SearchInfo["IPID"] = SearchInfo["IPID"].astype("Int64")
SearchInfo
```

```python
SearchStream = pd.read_parquet(f"{dataset_path}/SearchStream.pqt")
SearchStream["SearchStreamID"] = np.arange(len(SearchStream))
for col in ["Position", "ObjectType", "IsClick"]:
    SearchStream[col] = SearchStream[col].astype("Int64")
SearchStream
```

```python
PhoneRequestsStream = pd.read_parquet(f"{dataset_path}/PhoneRequestsStream.pqt")
PhoneRequestsStream["PhoneRequestID"] = np.arange(len(PhoneRequestsStream))
PhoneRequestsStream["IPID"] = PhoneRequestsStream["IPID"].astype("Int64")
PhoneRequestsStream
```

```python
UserInfo = pd.read_parquet(f"{dataset_path}/UserInfo.pqt")
for col in ["UserID", "UserAgentID", "UserAgentOSID", "UserAgentFamilyID", "UserDeviceID"]:
    UserInfo[col] = UserInfo[col].astype("Int64")
    # Add turn nan to MaxID + 1
    UserInfo[col] = UserInfo[col].fillna(UserInfo[col].max() + 1)
UserInfo
```

```python
VisitStream = pd.read_parquet(f"{dataset_path}/VisitStream.pqt")
VisitStream["VisitID"] = np.arange(len(VisitStream))
VisitStream["IPID"] = VisitStream["IPID"].astype("Int64")
VisitStream
```

```python
user_visits_train = pd.read_parquet(f"{dataset_path}/user-visits/train.pqt")
user_visits_test = pd.read_parquet(f"{dataset_path}/user-visits/test.pqt")
user_visits_validation = pd.read_parquet(f"{dataset_path}/user-visits/validation.pqt")
```

```python
# change the column num_click to num_visits
user_visits_train.rename(columns={"num_click": "num_visits"}, inplace=True)
user_visits_test.rename(columns={"num_click": "num_visits"}, inplace=True)
user_visits_validation.rename(columns={"num_click": "num_visits"}, inplace=True)

user_visits_train
```

```python
# Save the parquet files under griffin_dataset
import shutil
import os
shutil.rmtree(griffin_path, ignore_errors=True)
shutil.copytree(dataset_path, griffin_path)
```

```python
SearchStream.to_parquet(f"{griffin_path}/SearchStream.pqt")
PhoneRequestsStream.to_parquet(f"{griffin_path}/PhoneRequestsStream.pqt")
VisitStream.to_parquet(f"{griffin_path}/VisitStream.pqt")
UserInfo.to_parquet(f"{griffin_path}/UserInfo.pqt")
location.to_parquet(f"{griffin_path}/Location.pqt")
SearchInfo.to_parquet(f"{griffin_path}/SearchInfo.pqt")

user_visits_train.to_parquet(f"{griffin_path}/user-visits/train.pqt")
user_visits_test.to_parquet(f"{griffin_path}/user-visits/test.pqt")
user_visits_validation.to_parquet(f"{griffin_path}/user-visits/validation.pqt")

```
