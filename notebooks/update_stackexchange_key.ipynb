```python
import os
from pathlib import Path
import shutil
import pandas as pd
import numpy as np

data_dir = Path('../datasets/stackexchange')
griffin_data_dir = Path('../griffin_datasets/stackexchange-pk')
```

```python
# read the postTags table
postTags = pd.read_parquet(Path(data_dir) / 'data' / 'postTags.pqt')
postTags
```

```python
# Add a new column 'post_tag_id' to the postTags table
postTags['post_tag_id'] = np.arange(len(postTags))
postTags
```

```python
# Save the whole datasets to the griffin_data_dir
shutil.copytree(Path(data_dir), Path(griffin_data_dir))

# Save the postTags table to the griffin_data_dir
postTags.to_parquet(Path(griffin_data_dir) / 'data' / 'postTags.pqt')
```
