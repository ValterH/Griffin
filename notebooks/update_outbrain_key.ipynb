```python
# For other tables, we need to add a primary key column
PageView = pd.read_parquet(Path(data_dir) / 'data' / 'page_views.pqt')
PageView
```

```python
pageview_id = np.arange(len(PageView))
pageview_id_map = pd.Series(pageview_id, index=PageView.index)
PageView['pageview_id'] = pageview_id_map
PageView
```

```python
DocumentsTopic = pd.read_parquet(Path(data_dir) / 'data' / 'documents_topics.pqt')
DocumentsTopic
```

```python
documents_topic_id = np.arange(len(DocumentsTopic))
documents_topic_id_map = pd.Series(documents_topic_id, index=DocumentsTopic.index)
DocumentsTopic['documents_topic_id'] = documents_topic_id_map
DocumentsTopic
```

```python
DocumentsCategory = pd.read_parquet(Path(data_dir) / 'data' / 'documents_categories.pqt')
DocumentsCategory
documents_category_id = np.arange(len(DocumentsCategory))
documents_category_id_map = pd.Series(documents_category_id, index=DocumentsCategory.index)
DocumentsCategory['documents_category_id'] = documents_category_id_map
DocumentsCategory

DocumentsEntity = pd.read_parquet(Path(data_dir) / 'data' / 'documents_entities.pqt')
DocumentsEntity
documents_entity_id = np.arange(len(DocumentsEntity))
documents_entity_id_map = pd.Series(documents_entity_id, index=DocumentsEntity.index)
DocumentsEntity['documents_entity_id'] = documents_entity_id_map
DocumentsEntity
```

```python
# copy the whole folder to outbrain-small-griffin
import shutil
import os
# shutil.copytree(Path('/home/ubuntu/Griffin_pure/griffin_datasets/outbrain-small'), Path('/home/ubuntu/Griffin_pure/griffin_datasets/outbrain-small-griffin'))
if os.path.exists(Path(griffin_data_dir)):
    shutil.rmtree(Path(griffin_data_dir))
shutil.copytree(Path(data_dir), Path(griffin_data_dir))
```

```python
# change the click table in the new folder
# remove the old click table
shutil.rmtree(Path(griffin_data_dir) / 'ctr')
# create a new click table
Path(Path(griffin_data_dir) / 'ctr').mkdir(parents=True, exist_ok=True)
click_train.to_parquet(Path(griffin_data_dir) / 'ctr' / 'train.pqt')
click_validation.to_parquet(Path(griffin_data_dir) / 'ctr' / 'validation.pqt')
click_test.to_parquet(Path(griffin_data_dir) / 'ctr' / 'test.pqt')

```

```python
# remove the old data file click.pqt
shutil.rmtree(Path(griffin_data_dir) / 'data' / 'clicks.pqt')
# add the new click table to the data folder
data.to_parquet(Path(griffin_data_dir) / 'data' / 'clicks.pqt')
```

```python
# Save other tables
shutil.rmtree(Path(griffin_data_dir) / 'data' / 'page_views.pqt')
PageView.to_parquet(Path(griffin_data_dir) / 'data' / 'page_views.pqt')
shutil.rmtree(Path(griffin_data_dir) / 'data' / 'documents_topics.pqt')
DocumentsTopic.to_parquet(Path(griffin_data_dir) / 'data' / 'documents_topics.pqt')
shutil.rmtree(Path(griffin_data_dir) / 'data' / 'documents_categories.pqt')
DocumentsCategory.to_parquet(Path(griffin_data_dir) / 'data' / 'documents_categories.pqt')
shutil.rmtree(Path(griffin_data_dir) / 'data' / 'documents_entities.pqt')
DocumentsEntity.to_parquet(Path(griffin_data_dir) / 'data' / 'documents_entities.pqt')
```
