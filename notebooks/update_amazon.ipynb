```python
# save to parquet
# copy the whold dataset to griffin_datasets
shutil.rmtree(griffin_path, ignore_errors=True)
shutil.copytree(dataset_path, griffin_path)

# save to parquet
shutil.rmtree(os.path.join(griffin_path, "data", "review.pqt"), ignore_errors=False)
shutil.rmtree(os.path.join(griffin_path, "rating"), ignore_errors=False)
os.makedirs(os.path.join(griffin_path, "rating"), exist_ok=True)
review.to_parquet(os.path.join(griffin_path, "data/review.pqt"))
rating_train.to_parquet(os.path.join(griffin_path, "rating/train.pqt"))
rating_validation.to_parquet(os.path.join(griffin_path, "rating/validation.pqt"))
rating_test.to_parquet(os.path.join(griffin_path, "rating/test.pqt"))
```
