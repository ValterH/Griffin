```python
import numpy as np
import pandas as pd
import os
import shutil
```

```python
dataset_path = "../datasets/avs"
griffin_path = "../griffin_datasets/avs-pk"
```

```python
history = pd.read_parquet(os.path.join(dataset_path, "data", "history.pqt"))
```

```python
history
```

```python
history['history_id'] = history.index
```

```python
history
```

```python
repeater_train = pd.read_parquet(os.path.join(dataset_path, "repeater", "train.pqt"))
repeater_validation = pd.read_parquet(os.path.join(dataset_path, "repeater", "validation.pqt"))
repeater_test = pd.read_parquet(os.path.join(dataset_path, "repeater", "test.pqt"))
```

```python
repeater_train
```

```python
repeater_train['history_id'] = repeater_train.index
repeater_validation['history_id'] = repeater_validation.index
repeater_test['history_id'] = repeater_test.index

```

```python
repeater_test
```

```python
# downsample the transactions table
transactions = pd.read_parquet(os.path.join(dataset_path, "data", "transactions.pqt"))
transactions = transactions.sample(frac=0.1, random_state=42)
transactions

```

```python
# Add a primary key to the transactions table
transactions['transaction_id'] = np.arange(len(transactions))
transactions
```

```python
# copy the data to the griffin dataset
shutil.rmtree(griffin_path, ignore_errors=True)
shutil.copytree(dataset_path, griffin_path)

# remove the history table
shutil.rmtree(os.path.join(griffin_path, "data", "history.pqt"), ignore_errors=False)
history.to_parquet(os.path.join(griffin_path, "data", "history.pqt"))

# remove the transactions table
shutil.rmtree(os.path.join(griffin_path, "data", "transactions.pqt"), ignore_errors=False)
transactions.to_parquet(os.path.join(griffin_path, "data", "transactions.pqt"))

shutil.rmtree(os.path.join(griffin_path, "repeater"), ignore_errors=False)
os.mkdir(os.path.join(griffin_path, "repeater"))

repeater_train.to_parquet(os.path.join(griffin_path, "repeater", "train.pqt"))
repeater_validation.to_parquet(os.path.join(griffin_path, "repeater", "validation.pqt"))
repeater_test.to_parquet(os.path.join(griffin_path, "repeater", "test.pqt"))
```
