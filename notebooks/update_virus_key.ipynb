```python
import os
from pathlib import Path
import shutil
import pandas as pd
import numpy as np

data_dir = Path('../griffin_datasets/virus-griffin-v3')
griffin_data_dir = Path('../griffin_datasets/virus-pk')
```

```python
weather = pd.read_parquet(Path(data_dir) / 'weather.pqt')
spray = pd.read_parquet(Path(data_dir) / 'spray.pqt')
```

```python
# Add a new column 'weather_id' to the weather table
weather['weather_id'] = np.arange(len(weather))

# Add a new column 'spray_id' to the spray table
spray['spray_id'] = np.arange(len(spray))
```

```python
spray
```

```python
# Save the whole datasets to the griffin_data_dir
shutil.copytree(Path(data_dir), Path(griffin_data_dir))

# Save the weather and spray tables to the griffin_data_dir
weather.to_parquet(Path(griffin_data_dir) / 'weather.pqt')
spray.to_parquet(Path(griffin_data_dir) / 'spray.pqt')
```
